Objective: combine missing-value handling, duplicate removal, and data-type correction.

```python
import pandas as pd
df = pd.read_csv('student_data_unclean.csv', na_values=['missing'])
print('Shape:', df.shape)
display(df.head())
```

## 1. Inspect data

```python
print(df.dtypes)
print('Missing values:\n', df.isnull().sum())
print('Duplicate rows:', df.duplicated().sum())
```

## 2. Correct data types

```python
df['age'] = pd.to_numeric(df['age'], errors='coerce')
df['score'] = pd.to_numeric(df['score'], errors='coerce')
print(df.dtypes)
```

## 3. Remove duplicates

```python
before = len(df)
df = df.drop_duplicates()
print('Before:', before, 'After:', len(df), 'Removed:', before-len(df))
```

## 4. Handle missing values

```python
age_median = df['age'].median()
score_median = df['score'].median()
df['age'] = df['age'].fillna(age_median)
df['score'] = df['score'].fillna(score_median)
print('Medians used:', age_median, score_median)
print(df.isnull().sum())
```

## 5. Validate and save

```python
print('Final shape:', df.shape)
print('Remaining duplicates:', df.duplicated().sum())
print(df.dtypes)
df.to_csv('student_data_cleaned.csv', index=False)
print('Saved student_data_cleaned.csv')
```

## Cleaning decisions
- Blank ages and the score text `missing` are treated as missing values.
- Age and score are converted to numeric with `pd.to_numeric()`.
- Exact duplicate rows are removed with `drop_duplicates()`.
- Numerical missing values are filled with their median, which is less sensitive to outliers than the mean.
- The final dataset is validated and exported.
